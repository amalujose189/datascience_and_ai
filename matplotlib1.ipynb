```python
import matplotlib.pyplot as plt
```

```python
x=[1,2,3,4,5]
y=[5,9,2,7,10]
plt.plot(x,y)
plt.xlabel("Days")
plt.ylabel("Sales")
plt.title("Daily Sales")
plt.show()
```

```python
x=[1,2,3,4,5]
plt.subplot(1,2,1)
plt.plot(x,[i*i for i in x]) #squares
plt.subplot(1,2,2)
plt.plot(x,[i*i*i for i in x])
plt.show()#cubes
```

```python
x=[10,20,30,40,50]
y=[15,25,35,30,45]
plt.scatter(x,y)
plt.xlabel("Age")
plt.ylabel("income")
plt.title("age vs income")
plt.show()
```

```python
x=[10,20,30,40,50]
y=[15,25,35,30,45]
sizes=[100,200,300,250,150]
plt.scatter(x,y,s=sizes,alpha=0.5)
plt.title("Bubble chart")
plt.show()
```

```python
days=["mon","tue","wed","thu","fri"]
sales=[120,100,160,140,200]
plt.bar(days,sales)
plt.title("sales by day")
plt.show()
```

```python
marks=[10,20,30,40,20,30,10,50,60,70,20,30,40]
plt.hist(marks,bins=5)
plt.title("student marks distribution")
plt.show()
```

```python
categories=['A','B','C','D']
values=[30,20,25,25]
plt.pie(values,labels=categories,autopct="%0.1f%%")
plt.show()
```
