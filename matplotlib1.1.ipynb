```python
import matplotlib.pyplot as plt
```

```python
x=[1,2,3,4,5]
y=[5,9,2,7,10]
plt.plot(x,y)
plt.xlabel("Days")
plt.ylabel("Sales")
plt.title("Daily Sales")
plt.show()
```

```python
x=[1,2,3,4,5]
plt.subplot(1,2,1)
plt.plot(x,[i*i for i in x]) #squares
plt.subplot(1,2,2)
plt.plot(x,[i*i*i for i in x])
plt.show()#cubes
```

```python
x=[10,20,30,40,50]
y=[15,25,35,30,45]
plt.scatter(x,y)
plt.xlabel("Age")
plt.ylabel("income")
plt.title("age vs income")
plt.show()
```

```python
x=[10,20,30,40,50]
y=[15,25,35,30,45]
sizes=[100,200,300,250,150]
plt.scatter(x,y,s=sizes,alpha=0.5)
plt.title("Bubble chart")
plt.show()
```

```python
days=["mon","tue","wed","thu","fri"]
sales=[120,100,160,140,200]
plt.bar(days,sales)
plt.title("sales by day")
plt.show()
```

```python
marks=[10,20,30,40,20,30,10,50,60,70,20,30,40]
plt.hist(marks,bins=5)
plt.title("student marks distribution")
plt.show()
```

```python
categories=['A','B','C','D']
values=[30,20,25,25]
plt.pie(values,labels=categories,autopct="%0.1f%%")
plt.show()
```

```python
import pandas as pd
```

```python
df=pd.read_csv("iris.csv")
print(df)

```

```python
plt.plot(df.index,df['SepalLengthCm'])
plt.title("Sepal Length Trend")
plt.xlabel("index")
plt.ylabel("Sepal Length")
plt.show()
```

```python
x=df.index
plt.subplot(1,2,1)
plt.plot(x,df['SepalWidthCm'])
plt.subplot(1,2,2)
plt.plot(x,df['PetalLengthCm'])
plt.show()
```

```python
plt.scatter(df['SepalLengthCm'],df['SepalWidthCm'])
plt.xlabel("sepal length")
plt.ylabel("sepal width")
plt.title("sepal scatter")
plt.show()
#to check overall spread
```

```python
plt.scatter(df['SepalLengthCm'],df['SepalWidthCm'],s=df['PetalLengthCm']*50,alpha=0.5)#*50 because of small value ,alpha for transparency
plt.title("bubble chart")
plt.ylabel("SepalWidthCm")
plt.show()
```

```python
df.groupby("Species")["SepalLengthCm"].mean().plot(kind='bar')
plt.xlabel("Species")
plt.ylabel("Mean Sepal Length")
plt.title("Avg Sepal Length by Species")
plt.show()
```

```python
plt.hist(df['PetalLengthCm'],bins=5)#convert into 5 intervals
plt.title("petal length distribution")
plt.xlabel("petal length")
plt.show()
```

```python
df['Species'].value_counts().plot(kind='pie',autopct='%0.1f%%')
plt.title('species percentage')
plt.ylabel("")
plt.show()#to show percentage
#for porportional comparison
```

```python
df.iloc[:,1:5].boxplot()# : means all row values should consider ,boxplotis used to check the outliers
plt.title("feature boxplot")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
iris=sns.load_dataset("iris")
#seaborn build above the matplotlib.seaborn has no show()
```

```python
iris
```

```python
sns.scatterplot(
    x="sepal_length",
    y="sepal_width",
    hue="species",
    data=iris
)
plt.show()
```

```python
sns.lineplot(
    x="sepal_length",
    y="petal_length",
    data=iris
)
plt.show()
```

```python
sns.barplot(
    x="species",
    y="petal_length",
    data=iris
)
plt.show()
```

```python
sns.histplot(
    iris["sepal_length"],
    bins=15
)
plt.show()
```
